```python
# test_example.py
import numpy as np
from scipy.io import loadmat
from sklearn.preprocessing import MinMaxScaler
from WNCROD import WNCROD   # نسخه‌ی اصلی نویسنده
```

```python

def test_example():
    # 1) بارگذاری مثال متلب
    mat = loadmat('Example.mat')      # مطمئن شو Example.mat کنار این فایل باشد
    trandata = mat['Example']         # مطابق README

    # 2) همان کاری که در README نوشته‌اند:
    #    در متلب normalize روی ستون‌های 2‌:3 انجام شده، اما در پایتون
    #    معمولا همه‌ی ستون‌های عددی را [0,1] می‌کنیم:
    trandata[:,1:] = MinMaxScaler().fit_transform(trandata[:,1:])

    # 3) X_tem = [0,1,4,5] یا [1,2,5,6] بسته به ایندکس متلب (یک مبنا)
    #    در پایتون اگر بخواهیم ایندکس صفر-مبنا، احتمالا:
    X_tem = [0, 1, 4, 5]
    lammda = 1.0

    scores = WNCROD(trandata, X_tem, lammda)
    print("out_scores =", np.round(scores, 5))

if __name__ == '__main__':
    test_example()
```
